```python
from ultralytics import YOLO

# 1. Загружаем предобученную на COCO модель YOLOv8 Nano
model = YOLO('yolov8n.pt')

# 2. Стартуем обучение
results = model.train(
    data='/Users/alexander/Developer/pcbcv/dataset/data.yaml',
    epochs=50,              # Достаточно для оценки сходимости
    batch=8,                # Безопасный размер для 16ГБ RAM
    imgsz=640,              # Стандартный размер, с которым YOLO работает лучше всего
    device='cpu',           # Твой надёжный выбор, чтобы исключить проблемы с MPS
    workers=4,
    lr0=0.01,               # Стандартная настройка
    momentum=0.937,
    weight_decay=0.0005,
    augment=True,           # Базовая аугментация YOLO включена
    project='pcb_defect_baseline',
    name='baseline_run',
    exist_ok=True,
    verbose=True
)
```
